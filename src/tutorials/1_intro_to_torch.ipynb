## Basics

```python
# Tensor
data = [[1, 2, 3], [4, 5, 6]]
my_tensor = torch.tensor(data)
print(my_tensor)

# Shapes
shape = (2, 3)
ones = torch.ones(shape)
zeros = torch.zeros(shape)
random = torch.randn(shape)
print(f"Random Tensor:\n {random}")

# Like and tensor properties
random_like = torch.rand_like(random)
print(f"Random Tensor:\n {random_like}")
print(f"Shape: {random_like.shape}")
print(f"Type: {random_like.dtype}")
print(f"Device: {random_like.device}")

# Data vs. Parameter
x_data = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
w = torch.tensor([[1.0], [2.0]], requires_grad=True)
print(f"Data tensor requires_grad: {x_data.requires_grad}")
print(f"Weights tensor requires_grad: {w.requires_grad}")

```

```python
# Compute z=x.y
a = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(3.0, requires_grad=True)
x = torch.tensor(4.0, requires_grad=True)

y = a + b  # First operation
z = x * y  # Second operation
print(f"Result z: {z}")

# Graph structure
print(f"grad_fn for z: {z.grad_fn}")  # Created by multiplication
print(f"grad_fn for y: {y.grad_fn}")  # Created by add
print(f"grad_fn for a: {a.grad_fn}")  # User created (None)
```

```python
# Element-wise and Matrix-wise multiplication 

a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[10, 20], [30, 40]])

element_wise_product = a * b
print(element_wise_product)

c = torch.tensor([[100, 200, 300, 400], [500, 600, 700, 800]])
mat_mul = a @ c
print(mat_mul)
```

```python
# Reductions
rnd = torch.randn((2,5))
print(f"Mean in first dimension (0th-index): {rnd.mean(dim=0)}")
print(f"Mean in second dimension (1st-index): {rnd.mean(dim=1)}")
```

```python
# Basic Indexing
x = torch.arange(12).reshape(3, 4)
print(f"x col 2: {x[:,2]}")

# Dynamic Argmax
scores = torch.tensor([
    [10, 0, 5, 20, 1],  # Best score at index 3
    [1, 30, 2, 5, 0]    # Best score at incex 2
])

best_indices = torch.argmax(scores, dim=1)
print(best_indices)
```

```python
# Gather (fast)
data = torch.tensor([
    [11, 12, 13, 14],    # row 0
    [21, 22, 23, 24],    # row 1
    [31, 32, 33, 34],    # row 2
])

indices_to_select = torch.tensor([
    [2],[0],[3]
])

selected_values = torch.gather(data, dim=1, index=indices_to_select)
print(selected_values)
```
